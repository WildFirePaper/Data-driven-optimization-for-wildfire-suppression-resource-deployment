```python
import gurobipy as gp
from gurobipy import GRB
import numpy as np
from math import exp, factorial
from scipy.stats import poisson
import pandas as pd
from itertools import product
from math import radians, cos, sin, sqrt, atan2
from scipy.sparse import lil_matrix
from scipy.sparse import identity
from scipy.sparse import dok_matrix
from scipy.sparse import csr_matrix

class TransientProbClass:
    def __init__(self, s, k, K, time, lamb, mu, p_0=None):
        self.s = s
        self.k = k
        self.K = K
        self.lamb = lamb
        self.mu = mu
        self.time = time
        self.states = self.generate_states_with_strict_constraint()
        self.num_states = len(self.states)
        self.p_0 = p_0 if p_0 is not None else self.create_initial_probability()
        self.Q = self.create_q_matrix_corrected()

    def generate_states_with_strict_constraint(self):
        states = []
        for queue_size in range(self.K + 1):
            for phase_distribution in product(range(self.s + 1), repeat=self.k):
                if sum(phase_distribution) > self.s or sum(phase_distribution) + queue_size > self.s + self.K:
                    continue
                if sum(phase_distribution) < self.s and queue_size > 0:
                    continue
                states.append((queue_size,) + tuple(phase_distribution))
        return states

    def create_initial_probability(self):
        row = np.zeros(self.num_states)
        row[0] = 1
        return row

    def create_q_matrix_corrected(self):
        states = self.generate_states_with_strict_constraint()
        Q = lil_matrix((self.num_states, self.num_states))
        state_indices = {state: index for index, state in enumerate(states)}

        for state in states:
            state_index = state_indices[state]
            queue_size, *phase_distribution = state

            if self.k == 1:
                if queue_size < self.K:
                    if sum(phase_distribution) < self.s:
                        new_state = (queue_size, phase_distribution[0] + 1)
                    else:
                        new_state = (queue_size + 1, phase_distribution[0])
                    new_state_index = state_indices.get(new_state)
                    if new_state_index is not None:
                        Q[state_index, new_state_index] = self.lamb

                if phase_distribution[0] > 0:
                    if queue_size > 0:
                        new_state = (queue_size - 1, phase_distribution[0])
                    else:
                        new_state = (queue_size, phase_distribution[0] - 1)
                    new_state_index = state_indices.get(new_state)
                    if new_state_index is not None:
                        Q[state_index, new_state_index] = self.mu * phase_distribution[0]

            else:
                if queue_size < self.K:
                    if sum(phase_distribution) < self.s:
                        new_state = (queue_size, phase_distribution[0] + 1) + tuple(phase_distribution[1:])
                    else:
                        new_state = (queue_size + 1,) + tuple(phase_distribution)
                    new_state_index = state_indices.get(new_state)
                    if new_state_index is not None:
                        Q[state_index, new_state_index] = self.lamb

                for phase in range(1, self.k + 1):
                    if phase_distribution[phase - 1] > 0:
                        if phase < self.k:
                            new_phase_dist = list(phase_distribution)
                            new_phase_dist[phase - 1] -= 1
                            new_phase_dist[phase] += 1
                            new_state = (queue_size,) + tuple(new_phase_dist)
                        else:
                            if queue_size > 0:
                                new_state = (queue_size - 1, phase_distribution[0] + 1) + tuple(phase_distribution[1:-1]) + (phase_distribution[-1] - 1,)
                            else:
                                new_state = (queue_size,) + tuple(phase_distribution[:-1]) + (phase_distribution[-1] - 1,)
                        new_state_index = state_indices.get(new_state)
                        if new_state_index is not None:
                            Q[state_index, new_state_index] = self.mu * phase_distribution[phase - 1]

        for i in range(self.num_states):
            Q[i, i] = -np.sum(Q[i, :])

        #new for sparsity
        Q = Q.tocsr()

        return Q

    def compute_max_rate(self):
        return max(-self.Q.diagonal())


    def compute_p_tilde(self):
      max_rate = self.compute_max_rate()
      if max_rate == 0:
          return identity(self.num_states).tocsr()
      return (self.Q.multiply(1 / max_rate) + identity(self.num_states)).tocsr()


    def check_poisson_sum(self, max_value):
        sum_poisson = 0
        m = 0
        error_tolerance = 10**(-4)
        while True:
            probability = poisson.pmf(m, max_value * self.time)
            sum_poisson += probability
            if sum_poisson >= 1 - error_tolerance:
                break
            m += 1
        return sum_poisson, m + 1


def setup_optimization_with_intervals(k, K, time_intervals, lamb_intervals, mu_intervals, max_v, thresholds, num_areas, initial_resources, province_supply_limit, distance_df, cost_y, gamma, arrival_intervals):
    model = gp.Model("Resource Optimization with Randomization for Multiple Areas")

    # Create an environment with your WLS license
    params = {
    "WLSACCESSID": '----',
    "WLSSECRET": '----',
    "LICENSEID": ---,
    }

    env = gp.Env(params=params)
    # Create the model1 within the Gurobi environment
    model = gp.Model(env=env)

    # Add solver parameters
    model.setParam('Method', -1) #-1
    model.setParam('Presolve', 2)
    model.setParam('Cuts', 2)
    model.setParam('MIPGap', 0.001)
    model.setParam('Threads', 12)
    model.setParam('PreSparsify', 1)    # Automatically sparsify the presolve step

    # Create decision variables
    z_vars = model.addVars(num_areas, range(1, max_v + 1), vtype=GRB.BINARY, name="z")
    y_vars = model.addVars(num_areas, vtype=GRB.INTEGER, name="y")
    x_vars = model.addVars(num_areas, num_areas, vtype=GRB.INTEGER, name="x")

    # Servers allocated
    s_vars = {j: gp.quicksum(v * z_vars[j, v] for v in range(1, max_v + 1)) for j in range(num_areas)}

    # Waiting time variables
    tot_wait_vars = model.addVars(num_areas, vtype=GRB.CONTINUOUS, name="tot_wait")

    # Objective function components
    cost_allocation = gamma * gp.quicksum(s_vars[j] for j in range(num_areas))
    cost_transportation = gp.quicksum(distance_df.iloc[j, j_prime] * x_vars[j, j_prime] for j in range(num_areas) for j_prime in range(num_areas) if j != j_prime)
    cost_supply = gp.quicksum(cost_y[j] * y_vars[j] for j in range(num_areas))


    model.setObjective(cost_allocation + cost_transportation + cost_supply, GRB.MINIMIZE)

    prob_mass_store = {}

    # Constraints
    for j in range(num_areas):
        # Resource balance constraint
        model.addConstr(
            s_vars[j] == initial_resources[j] + y_vars[j] +
            gp.quicksum(x_vars[j_prime, j] for j_prime in range(num_areas) if j_prime != j)
            - gp.quicksum(x_vars[j, j_prime] for j_prime in range(num_areas) if j_prime != j),
            name=f"resource_balance_{j}"
        )

        # Server selection constraint
        model.addConstr(
            gp.quicksum(z_vars[j, v] for v in range(1, max_v + 1)) == 1,
            name=f"one_server_{j}"
        )

        # Time interval-based transient probability updates
        phi_vars = {}
        p_vars = {}
        tot_wait_j = 0


        for v in range(1, max_v + 1):

            previous_pi_t = None

            for interval in range(len(time_intervals)):

                time = time_intervals[interval]
                lamb_v = lamb_intervals[j][interval][v]
                mu_v = mu_intervals[j][interval][v]

                transient = TransientProbClass(v, k, K, time, lamb_v, mu_v)

                phi_0 = transient.p_0
                P_tilde = transient.compute_p_tilde()
                states = transient.states
                max_rate = transient.compute_max_rate()

                _, m = transient.check_poisson_sum(max_rate)

                p_vars[(v, interval)] = poisson.pmf(
                    range(m + 1),
                    mu=max_rate * time
                )

                # Start-of-interval probability vector
                phi_vars[(v, interval, 0)] = model.addVars(
                    len(states),
                    vtype=GRB.CONTINUOUS,
                    lb=0,
                    ub=1,
                    name=f"phi_{j}_{v}_{interval}_0"
                )


                if interval == 0:
                    model.addConstrs(
                        (
                            phi_vars[(v, interval, 0)][i] == phi_0[i] * z_vars[j, v]
                            for i in range(len(states))
                        ),
                        name=f"phi_init_{j}_{v}_{interval}"
                    )
                else:
                    model.addConstrs(
                        (
                            phi_vars[(v, interval, 0)][i] == previous_pi_t[i]
                            for i in range(len(states))
                        ),
                        name=f"phi_link_pi_{j}_{v}_{interval}"
                    )

                # Uniformization recursion
                for n in range(1, m + 1):

                    phi_vars[(v, interval, n)] = model.addVars(
                        len(states),
                        vtype=GRB.CONTINUOUS,
                        lb=0,
                        ub=1,
                        name=f"phi_{j}_{v}_{interval}_{n}"
                    )

                    for i in range(len(states)):
                        incoming_states = P_tilde[:, i].nonzero()[0]

                        model.addConstr(
                            phi_vars[(v, interval, n)][i]
                            ==
                            gp.quicksum(
                                P_tilde[k_state, i]
                                * phi_vars[(v, interval, n - 1)][k_state]
                                for k_state in incoming_states
                            ),
                            name=f"phi_rec_{j}_{v}_{interval}_{n}_{i}"
                        )

                # Correct transient distribution at end of interval
                pi_t_vars = [
                    gp.quicksum(
                        phi_vars[(v, interval, n)][i] * p_vars[(v, interval)][n]
                        for n in range(m + 1)
                    )
                    for i in range(len(states))
                ]


                # Use pi_t_vars, not phi_m, for next interval
                previous_pi_t = pi_t_vars

                # Expected queue length at end of this interval
                E_Lq = gp.quicksum(
                    pi_t_vars[i] * states[i][0]
                    for i in range(len(states))
                )

                # TEW only until T
                #if interval < arrival_intervals:
                tot_wait_j += E_Lq * time

        # Assign total waiting time for area j
        model.addConstr(tot_wait_vars[j] == tot_wait_j, name=f"total_wait_{j}")

        # Waiting time threshold constraint
        model.addConstr(tot_wait_vars[j] <= thresholds[j], name=f"wait_threshold_{j}")


    # Province supply limit constraint
    model.addConstr(gp.quicksum(y_vars[j] for j in range(num_areas)) <= province_supply_limit, name="province_supply_limit")

    model.optimize()
    print("Gurobi status:", model.status)

    if model.status == GRB.INFEASIBLE:
      model.computeIIS()
      model.write("infeasible_model.ilp")
      print("IIS written to infeasible_model.ilp")


    # After solving the model
    if model.status == GRB.OPTIMAL:
        for j in range(num_areas):
            print(f"Area {j} - Final waiting time: {tot_wait_vars[j].x} compared to threshold {thresholds[j]}")

    # Retrieve results
    if model.status == GRB.OPTIMAL:
        optimal_s = {j: sum(v * z_vars[j, v].x for v in range(1, max_v + 1)) for j in range(num_areas)}
        optimal_x = {(j, j_prime): x_vars[j, j_prime].x for j in range(num_areas) for j_prime in range(num_areas) if j != j_prime}
        optimal_y = {j: y_vars[j].x for j in range(num_areas)}

        allocation_val = gamma * sum(sum(v * z_vars[j, v].x for v in range(1, max_v + 1)) for j in range(num_areas))
        transport_val = sum(distance_df.iloc[j, j_prime] * x_vars[j, j_prime].x
                            for j in range(num_areas)
                            for j_prime in range(num_areas)
                            if j != j_prime)
        supply_val = sum(cost_y[j] * y_vars[j].x for j in range(num_areas))
        total_objective_val = model.ObjVal

        total_distance_used = sum(
            distance_df.iloc[j, j_prime]
            for j in range(num_areas)
            for j_prime in range(num_areas)
            if j != j_prime and x_vars[j, j_prime].x > 0
        )

        if model.status == GRB.OPTIMAL:
          for j in range(num_areas):
              print(
                  f"Area {j} - TEW: {tot_wait_vars[j].x:.6f}, "
                  f"threshold: {thresholds[j]}"
              )

          transfer_rows = []

          for j in range(num_areas):
              for j_prime in range(num_areas):
                  if j != j_prime and x_vars[j, j_prime].x > 1e-6:
                      transfer_rows.append({
                          "from_area": AREA_NAMES[j],
                          "to_area": AREA_NAMES[j_prime],
                          "units_transferred": x_vars[j, j_prime].x,
                          "distance_km": distance_df.iloc[j, j_prime],
                      })

          transfer_df = pd.DataFrame(transfer_rows)
          transfer_df.to_csv("resource_transfers.csv", index=False)

          print("\nSaved transfers to resource_transfers.csv")
          print(transfer_df)

        if model.status != GRB.OPTIMAL:
            print(f"Model not optimal. Status: {model.status}")

        print(f"\nObjective Function Breakdown:")
        print(f"  Allocation Cost (γ × ∑s):       {allocation_val:.4f}")
        print(f"  Transportation Cost (∑x·d):     {transport_val:.4f}")
        print(f"  Supply Cost (∑y·c):             {supply_val:.4f}")
        print(f"  Total Objective:                {total_objective_val:.4f}")
        print(f"  Total Distance (∑1·d if x>0):   {total_distance_used:.4f}")

        print("\nResource Transfers and Distances (only x > 0):")
        for j in range(num_areas):
            for j_prime in range(num_areas):
                if j != j_prime and x_vars[j, j_prime].x > 0:
                    print(f"From Area {j} to Area {j_prime}: {x_vars[j, j_prime].x:.0f} units, Distance = {distance_df.iloc[j, j_prime]:.2f}")


        return optimal_s, optimal_x, optimal_y

    return None, None, None
```

```python
# Haversine formula to calculate distance between two coordinates
def haversine(coord1, coord2):
    R = 6371e3  # Radius of Earth in meters
    lat1, lon1 = coord1
    lat2, lon2 = coord2
    phi1, phi2 = radians(lat1), radians(lat2)
    delta_phi = radians(lat2 - lat1)
    delta_lambda = radians(lon2 - lon1)
    a = sin(delta_phi / 2)**2 + cos(phi1) * cos(phi2) * sin(delta_lambda / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    distance = (R * c)/1000
    return distance


# Create distance matrix
locations = {
    "FA-1": (53.5379791, -113.5072631),
    "FA-2": (56.727114, -111.381511),
    "FA-3": (55.2935138, -114.7676884),
    "FA-4": (56.236665, -117.295388),
    "FA-5": (54.7703547, -111.9603138),
    "FA-6": (55.1556118, -118.8214727),
    "FA-7": (53.5789151, -116.4371904),
    "FA-8": (52.3733848, -114.9204833),
    "FA-9": (54.1328792, -115.6588747),
    "FA-10": (51.1022345, -114.234016),
}


distance_matrix = np.zeros((len(locations), len(locations)))
location_names = list(locations.keys())
for i, loc1 in enumerate(location_names):
    for j, loc2 in enumerate(location_names):
        distance_matrix[i, j] = haversine(locations[loc1], locations[loc2]) if i != j else 0
distance_df = pd.DataFrame(distance_matrix, index=location_names, columns=location_names)
```

```python
# =========================
# AUTOMATE: per-date solves
# =========================

import os
import pandas as pd
import numpy as np
from decimal import Decimal, ROUND_HALF_UP

# =========================
# CONFIG
# =========================


AREA_NAMES = [
    "FA-1", "FA-2", "FA-3", "FA-4", "FA-5",
    "FA-6", "FA-7", "FA-8", "FA-9", "FA-10",
]

AREA_FILE_MAP = {
    area: f"Lambda_Mu_{area}.xlsx"
    for area in AREA_NAMES
}


THRESHOLDS          = [0.5] * 10
#INITIAL_RESOURCES   = [1] * 10
INITIAL_RESOURCES   = [1, 3, 4, 2, 1, 5, 3, 2, 3, 1]
PROVINCE_SUPPLY_LIM = 1000
COST_Y              = [1_000_000] * 10
GAMMA               = 1

K_PHASES = 2
K_BUFFER = 5
MAX_V    = 7

# Main fix:
INTERVAL_HOURS       = 1
ARRIVAL_CUTOFF_HOURS = 12     # T
TEW_HORIZON_HOURS    = 24     # T' -- test 14, 16, 18, 20, 22, 24

ARRIVAL_INTERVALS = int(ARRIVAL_CUTOFF_HOURS / INTERVAL_HOURS)
TOTAL_INTERVALS   = int(TEW_HORIZON_HOURS / INTERVAL_HOURS)

RESULTS_CSV = f"required_servers_T{ARRIVAL_CUTOFF_HOURS}_Tp{TEW_HORIZON_HOURS}.csv"

# =========================
# HELPERS
# =========================

def _find_col(df, candidates):
    norm = {str(c).strip().lower(): c for c in df.columns}
    for c in candidates:
        key = c.strip().lower()
        if key in norm:
            return norm[key]
    return None


def load_area_day_rates_smart(xlsx_path, start_row=None, end_row=None):
    df = pd.read_excel(xlsx_path)

    if start_row is None:
        start_idx = 0
    else:
        start_idx = max(start_row - 1, 0)

    if end_row is None:
        end_idx = len(df)
    else:
        end_idx = min(end_row, len(df))

    df = df.iloc[start_idx:end_idx, :]

    date_col = _find_col(df, ["FORECAST_DATE"])
    lam_col  = _find_col(df, ["lambda", "lam", "lamb", "λ"])
    mu_col   = _find_col(df, ["mu", "Mu", "service_rate"])

    if not (date_col and lam_col and mu_col):
        raise ValueError(f"Required columns not found in {xlsx_path}")

    tmp = df[[date_col, lam_col, mu_col]].copy()
    tmp.columns = ["date", "lambda", "mu"]

    tmp["date"] = pd.to_datetime(tmp["date"], errors="coerce").dt.date
    tmp = tmp.dropna(subset=["date"])

    tmp["lambda"] = pd.to_numeric(tmp["lambda"], errors="coerce")
    tmp["mu"]     = pd.to_numeric(tmp["mu"], errors="coerce")

    return tmp


def align_all_areas_to_union_dates(daily_by_area, mu_floor=None):
    all_dates = set()
    for df in daily_by_area.values():
        all_dates |= set(df["date"])

    full_dates = sorted(all_dates)
    full_df = pd.DataFrame({"date": full_dates})

    aligned = {}

    for area, df in daily_by_area.items():
        m = full_df.merge(df, on="date", how="left")

        m["lambda"] = pd.to_numeric(m["lambda"], errors="coerce").fillna(0.0).astype(float)
        m["mu"]     = pd.to_numeric(m["mu"], errors="coerce").fillna(0.0).astype(float)

        if mu_floor is not None:
            m.loc[m["mu"] == 0.0, "mu"] = float(mu_floor)

        aligned[area] = m

    return aligned, full_dates

# =========================
# KEY FIX: BUILD T AND T'
# =========================

def assemble_intervals_for_day(
    row,
    max_v,
    total_intervals,
    arrival_intervals,
    dt_hours=1
):
    """
    Creates interval-level lambda and mu values.

    T  = arrival cutoff time.
    T' = TEW computation horizon.

    For intervals before T:
        lambda = daily predicted lambda

    For intervals after T:
        lambda = 0

    Mu is kept unchanged because service continues after arrivals stop.
    """

    lam = float(row["lambda"])
    mu  = float(row["mu"])


    lamb_intervals = []
    mu_intervals   = []

    for interval in range(total_intervals):

        if interval < arrival_intervals:
            lam_now = lam
        else:
            lam_now = 0.0

        lamb_intervals.append({
            v: lam_now for v in range(1, max_v + 1)
        })

        mu_intervals.append({
            v: mu for v in range(1, max_v + 1)
        })

    time_intervals = [dt_hours] * total_intervals

    return lamb_intervals, mu_intervals, time_intervals

# =========================
# MAIN RUNNER
# =========================

def run_all_dates_and_save_required_servers(
    start_row=None,
    end_row=None,
    mu_floor=None,
    results_csv=None,
    tew_horizon_hours=TEW_HORIZON_HOURS
):
    """
    Runs the model for each date using:

    T  = ARRIVAL_CUTOFF_HOURS
    T' = tew_horizon_hours

    Arrivals occur only until T.
    TEW is computed until T'.
    """

    if results_csv is None:
        results_csv = f"required_servers_T{ARRIVAL_CUTOFF_HOURS}_Tp{tew_horizon_hours}.csv"

    total_intervals = int(tew_horizon_hours / INTERVAL_HOURS)
    arrival_intervals = int(ARRIVAL_CUTOFF_HOURS / INTERVAL_HOURS)

    if total_intervals <= arrival_intervals:
        raise ValueError("T' must be greater than T. Increase tew_horizon_hours.")

    print("=" * 80)
    print("RUN CONFIGURATION")
    print("=" * 80)
    print(f"Arrival cutoff T      : {ARRIVAL_CUTOFF_HOURS} hours")
    print(f"TEW horizon T'        : {tew_horizon_hours} hours")
    print(f"Interval length       : {INTERVAL_HOURS} hours")
    print(f"Arrival intervals     : {arrival_intervals}")
    print(f"Total TEW intervals   : {total_intervals}")
    print("=" * 80)

    daily_by_area = {
        area: load_area_day_rates_smart(
            AREA_FILE_MAP[area],
            start_row=start_row,
            end_row=end_row
        )
        for area in AREA_NAMES
    }

    aligned, dates = align_all_areas_to_union_dates(
        daily_by_area,
        mu_floor=mu_floor
    )

    all_rows = []
    infeasible_dates = []

    for d in dates:
        print(f"\nSolving date: {d}")

        lamb_all = []
        mu_all   = []
        time_intervals = [INTERVAL_HOURS] * total_intervals

        for area in AREA_NAMES:
            row = aligned[area].loc[aligned[area]["date"] == d].iloc[0]

            lamb_one, mu_one, _ = assemble_intervals_for_day(
                row=row,
                max_v=MAX_V,
                total_intervals=total_intervals,
                arrival_intervals=arrival_intervals,
                dt_hours=INTERVAL_HOURS
            )

            lamb_all.append(lamb_one)
            mu_all.append(mu_one)

        optimal_s, optimal_x, optimal_y = setup_optimization_with_intervals(
            k=K_PHASES,
            K=K_BUFFER,
            time_intervals=time_intervals,
            lamb_intervals=lamb_all,
            mu_intervals=mu_all,
            max_v=MAX_V,
            thresholds=THRESHOLDS,
            num_areas=len(AREA_NAMES),
            initial_resources=INITIAL_RESOURCES,
            province_supply_limit=PROVINCE_SUPPLY_LIM,
            distance_df=distance_df,
            cost_y=COST_Y,
            gamma=GAMMA,
            arrival_intervals=arrival_intervals,
        )

        if optimal_s is None:
            print(f"[INFEASIBLE] {d}")
            infeasible_dates.append(pd.Timestamp(d))
            continue

        for j, area in enumerate(AREA_NAMES):
            all_rows.append({
                "date": pd.Timestamp(d),
                "area": area,
                "required_servers": int(round(optimal_s[j])),
                "T_arrival_cutoff": ARRIVAL_CUTOFF_HOURS,
                "T_prime_tew_horizon": tew_horizon_hours
            })


    if infeasible_dates:
        pd.DataFrame({"date": infeasible_dates}).to_csv(
            f"infeasible_dates_Tp{tew_horizon_hours}.csv",
            index=False
        )

    results_df = pd.DataFrame(all_rows)

    if not results_df.empty:
        results_df = results_df.sort_values(["date", "area"])
        results_df.to_csv(results_csv, index=False)
        print(f"\nSaved results to: {results_csv}")
    else:
        print("\nNo feasible results to save.")

    return results_df
```

```python
# =========================
# TEST DIFFERENT T' VALUES
# =========================

def run_tew_horizon_sensitivity(
    horizons=(24),
    start_row=None,
    end_row=None,
    mu_floor=None
):
    sensitivity_outputs = {}
    summary_rows = []

    for Tp in horizons:
        print("\n" + "#" * 100)
        print(f"RUNNING T' = {Tp} HOURS")
        print("#" * 100)

        out = run_all_dates_and_save_required_servers(
            start_row=start_row,
            end_row=end_row,
            mu_floor=mu_floor,
            results_csv=f"required_servers_T{ARRIVAL_CUTOFF_HOURS}_Tp{Tp}.csv",
            tew_horizon_hours=Tp
        )

        sensitivity_outputs[Tp] = out

        if out is not None and not out.empty:
            summary_rows.append({
              "T_prime": Tp
          })

    summary_df = pd.DataFrame(summary_rows)
    summary_df.to_csv("tew_horizon_sensitivity_summary.csv", index=False)

    return sensitivity_outputs, summary_df
```

```python
import os
import numpy as np
import pandas as pd
from itertools import product
from scipy.stats import poisson
from scipy.special import factorial

# =========================
# CONFIG
# =========================

AREA_NAMES = [
    "FA-1", "FA-2", "FA-3", "FA-4", "FA-5",
    "FA-6", "FA-7", "FA-8", "FA-9", "FA-10",
]

AREA_FILE_MAP = {
    area: f"Lambda_Mu_{area}.xlsx"
    for area in AREA_NAMES
}

SHEET_NAME = "Full95"

K_PHASES = 2
QUEUE_CAPACITY = 5
MAX_SERVERS = 7

THRESHOLD = 0.5

T_ARRIVAL_CUTOFF = 12      # T
T_PRIME = 24               # T'
INTERVAL_HOURS = 1/12

OUTPUT_CSV = f"required_servers_no_optimization_T{T_ARRIVAL_CUTOFF}_Tp{T_PRIME}.csv"


# =========================
# TRANSIENT QUEUE CLASS
# =========================

class TransientProbClass:
    def __init__(self, s, k, K, total_time, lamb, mu, p_0=None):
        self.s = s
        self.k = k
        self.K = K
        self.lamb = lamb
        self.mu = mu
        self.total_time = total_time
        self.states = self.generate_states_with_strict_constraint()
        self.num_states = len(self.states)
        self.p_0 = p_0 if p_0 is not None else self.create_initial_probability()
        self.time_step = None

    def create_initial_probability(self):
        row = np.zeros(self.num_states)
        row[0] = 1.0
        return row

    def generate_states_with_strict_constraint(self):
        states = []
        for queue_size in range(self.K + 1):
            for phase_distribution in product(range(self.s + 1), repeat=self.k):
                if sum(phase_distribution) > self.s:
                    continue
                if sum(phase_distribution) + queue_size > self.s + self.K:
                    continue
                if sum(phase_distribution) < self.s and queue_size > 0:
                    continue
                states.append((queue_size,) + tuple(phase_distribution))
        return states

    def create_q_matrix_corrected(self):
        Q = np.zeros((self.num_states, self.num_states))
        state_indices = {state: idx for idx, state in enumerate(self.states)}

        for state in self.states:
            idx = state_indices[state]
            queue_size, *phase_distribution = state

            # Arrival
            if queue_size < self.K and self.lamb > 0:
                if sum(phase_distribution) < self.s:
                    new_state = (
                        queue_size,
                        phase_distribution[0] + 1
                    ) + tuple(phase_distribution[1:])
                else:
                    new_state = (queue_size + 1,) + tuple(phase_distribution)

                new_idx = state_indices.get(new_state)
                if new_idx is not None:
                    Q[idx, new_idx] = self.lamb

            # Service phase transitions
            for phase in range(1, self.k + 1):
                if phase_distribution[phase - 1] > 0:
                    if phase < self.k:
                        new_phase_dist = list(phase_distribution)
                        new_phase_dist[phase - 1] -= 1
                        new_phase_dist[phase] += 1
                        new_state = (queue_size,) + tuple(new_phase_dist)
                    else:
                        if queue_size > 0:
                            new_state = (
                                queue_size - 1,
                                phase_distribution[0] + 1
                            ) + tuple(phase_distribution[1:-1]) + (
                                phase_distribution[-1] - 1,
                            )
                        else:
                            new_state = (
                                queue_size,
                            ) + tuple(phase_distribution[:-1]) + (
                                phase_distribution[-1] - 1,
                            )

                    new_idx = state_indices.get(new_state)
                    if new_idx is not None:
                        Q[idx, new_idx] = self.mu * phase_distribution[phase - 1]

        for i in range(self.num_states):
            Q[i, i] = -np.sum(Q[i, :])

        return Q

    def calculate_metrics(self, probs):
        E_Lq = 0.0
        E_L = 0.0
        P_W = 0.0
        P_f = 0.0
        P_delay = 0.0

        for state, prob in zip(self.states, probs):
            queue = state[0]
            in_service = sum(state[1:])

            E_Lq += queue * prob
            E_L += (queue + in_service) * prob

            if queue > 0 or in_service == self.s:
                P_W += prob

            if queue == self.K:
                P_f += prob

            if queue > 0:
                P_delay += prob

        delayed_fires = self.lamb * P_delay * self.time_step

        return E_Lq, E_L, P_W, P_f, delayed_fires

    def estimate_transient_probabilities_and_metrics_with_cutoff(
        self,
        time_steps,
        arrival_cutoff_time
    ):
        self.time_step = self.total_time / time_steps

        phi = self.p_0.copy()
        total_waiting_time = 0.0
        cumulative_delayed_fires = 0.0

        original_lambda = self.lamb

        for step in range(time_steps):
            current_time = step * self.time_step

            if current_time < arrival_cutoff_time:
                self.lamb = original_lambda
            else:
                self.lamb = 0.0

            Q = self.create_q_matrix_corrected()
            max_rate = np.abs(np.diag(Q)).max()

            if max_rate == 0:
                pi_t = phi.copy()
            else:
                P_tilde = Q / max_rate + np.eye(self.num_states)

                sum_poisson = 0.0
                m = 0
                while True:
                    prob = poisson.pmf(m, max_rate * self.time_step)
                    sum_poisson += prob
                    if sum_poisson >= 1 - 1e-4:
                        break
                    m += 1

                poisson_probs = poisson.pmf(
                    np.arange(m + 1),
                    max_rate * self.time_step
                )

                phi_list = [phi]
                for n in range(1, m + 1):
                    phi_list.append(phi_list[-1].dot(P_tilde))

                pi_t = sum(phi_list[n] * poisson_probs[n] for n in range(m + 1))
                pi_t = pi_t / pi_t.sum()

            phi = pi_t

            E_Lq, E_L, P_W, P_f, delayed_fires = self.calculate_metrics(pi_t)

            total_waiting_time += E_Lq * self.time_step
            cumulative_delayed_fires += delayed_fires

        self.lamb = original_lambda

        return {
            "TEW": total_waiting_time,
            "final_E_L": E_L,
            "final_E_Lq": E_Lq,
            "final_P_W": P_W,
            "final_P_f": P_f,
            "total_delayed_fires": cumulative_delayed_fires,
        }

# =========================
# FILE HELPERS
# =========================

def _find_col(df, candidates):
    norm = {str(c).strip().lower(): c for c in df.columns}
    for c in candidates:
        key = c.strip().lower()
        if key in norm:
            return norm[key]
    return None


def load_area_day_rates_smart(xlsx_path, sheet_name="Full95", start_row=None, end_row=None):
    df = pd.read_excel(xlsx_path, sheet_name=sheet_name)

    if start_row is not None or end_row is not None:
        start_idx = 0 if start_row is None else max(start_row - 1, 0)
        end_idx = len(df) if end_row is None else min(end_row, len(df))
        df = df.iloc[start_idx:end_idx, :]

    date_col = _find_col(df, ["FORECAST_DATE"])
    lam_col = _find_col(df, ["lambda", "lam", "lamb", "λ"])
    mu_col = _find_col(df, ["mu", "Mu", "service_rate"])

    if not (date_col and lam_col and mu_col):
        raise ValueError(f"Could not find date/lambda/mu columns in {xlsx_path}")

    out = df[[date_col, lam_col, mu_col]].copy()
    out.columns = ["date", "lambda", "mu"]

    out["date"] = pd.to_datetime(out["date"], errors="coerce").dt.date
    out["lambda"] = pd.to_numeric(out["lambda"], errors="coerce")
    out["mu"] = pd.to_numeric(out["mu"], errors="coerce")

    out = out.dropna(subset=["date", "lambda", "mu"])

    return out


def align_all_areas_to_union_dates(daily_by_area, mu_floor=None):
    all_dates = set()

    for df in daily_by_area.values():
        all_dates |= set(df["date"])

    full_dates = sorted(all_dates)
    full_df = pd.DataFrame({"date": full_dates})

    aligned = {}

    for area, df in daily_by_area.items():
        merged = full_df.merge(df, on="date", how="left")

        merged["lambda"] = merged["lambda"].fillna(0.0).astype(float)
        merged["mu"] = merged["mu"].fillna(0.0).astype(float)

        if mu_floor is not None:
            merged.loc[merged["mu"] == 0.0, "mu"] = float(mu_floor)

        aligned[area] = merged

    return aligned, full_dates


# =========================
# SERVER SEARCH
# =========================

def find_min_servers_for_area(
    lam,
    mu,
    threshold,
    T,
    Tprime,
    interval_hours,
    max_servers,
    k_phases,
    queue_capacity
):
    time_steps = int(Tprime / interval_hours)

    if lam <= 0:
        return {
            "required_servers": 1,
            "TEW": 0.0,
            "final_E_L": 0.0,
            "final_E_Lq": 0.0,
            "feasible": True,
        }

    if mu <= 0:
        return {
            "required_servers": np.nan,
            "TEW": np.inf,
            "final_E_L": np.inf,
            "final_E_Lq": np.inf,
            "feasible": False,
        }

    last_metrics = None

    for s in range(1, max_servers + 1):
        transient = TransientProbClass(
            s=s,
            k=k_phases,
            K=queue_capacity,
            total_time=Tprime,
            lamb=lam,
            mu=mu
        )

        metrics = transient.estimate_transient_probabilities_and_metrics_with_cutoff(
            time_steps=time_steps,
            arrival_cutoff_time=T
        )

        last_metrics = metrics

        if metrics["TEW"] <= threshold:
            return {
                "required_servers": s,
                "TEW": metrics["TEW"],
                "final_E_L": metrics["final_E_L"],
                "final_E_Lq": metrics["final_E_Lq"],
                "feasible": True,
            }

    return {
        "required_servers": np.nan,
        "TEW": last_metrics["TEW"],
        "final_E_L": last_metrics["final_E_L"],
        "final_E_Lq": last_metrics["final_E_Lq"],
        "feasible": False,
    }
# =========================
# MAIN FUNCTION
# =========================

def compute_required_servers_no_optimization(
    area_file_map=AREA_FILE_MAP,
    area_names=AREA_NAMES,
    sheet_name=SHEET_NAME,
    threshold=THRESHOLD,
    T=T_ARRIVAL_CUTOFF,
    Tprime=T_PRIME,
    interval_hours=INTERVAL_HOURS,
    max_servers=MAX_SERVERS,
    k_phases=K_PHASES,
    queue_capacity=QUEUE_CAPACITY,
    start_row=None,
    end_row=None,
    mu_floor=None,
    output_csv=OUTPUT_CSV
):
    if Tprime <= T:
        raise ValueError("Tprime must be greater than T.")

    daily_by_area = {}

    for area in area_names:
        path = area_file_map[area]
        daily_by_area[area] = load_area_day_rates_smart(
            path,
            sheet_name=sheet_name,
            start_row=start_row,
            end_row=end_row
        )

    aligned, dates = align_all_areas_to_union_dates(
        daily_by_area,
        mu_floor=mu_floor
    )

    rows = []

    for d in dates:
        print(f"Processing date: {d}")

        for area in area_names:
            row = aligned[area].loc[aligned[area]["date"] == d].iloc[0]

            lam = float(row["lambda"])
            mu = float(row["mu"])

            result = find_min_servers_for_area(
                lam=lam,
                mu=mu,
                threshold=threshold,
                T=T,
                Tprime=Tprime,
                interval_hours=interval_hours,
                max_servers=max_servers,
                k_phases=k_phases,
                queue_capacity=queue_capacity
            )

            rows.append({
                "date": pd.Timestamp(d),
                "area": area,
                "lambda": lam,
                "mu": mu,
                "threshold": threshold,
                "T": T,
                "Tprime": Tprime,
                "required_servers": result["required_servers"],
                "TEW": result["TEW"],
                "final_E_L": result["final_E_L"],
                "final_E_Lq": result["final_E_Lq"],
                "feasible": result["feasible"],
            })

    results_df = pd.DataFrame(rows)
    results_df = results_df.sort_values(["date", "area"])

    results_df.to_csv(output_csv, index=False)
    print(f"Saved results to: {output_csv}")

    return results_df
```

# Lookup Table (Figure 9)

- This part will generate the look-up table (Figure 9) of the manuscript

```python
import numpy as np
import pandas as pd
from scipy.stats import poisson
from scipy.special import comb, factorial
from itertools import product

class TransientProbClass:
    def __init__(self, s, k, K, total_time, lamb, mu, p_0=None):
        self.s = s
        self.k = k
        self.K = K
        self.lamb = lamb
        self.mu = mu
        self.total_time = total_time  # Total time horizon
        self.num_states = self.calculate_number_of_states()
        self.time_step = 0  # Initialize time step, to be set in a loop
        self.p_0 = p_0 if p_0 is not None else self.create_initial_probability()

    def calculate_number_of_states(self):
        total_states = comb(self.s + self.k, self.s) + self.K * comb(self.s + self.k - 1, self.s)
        return int(total_states)

    def check_poisson_sum(self, max_value):
        sum_poisson = 0
        m = 0
        error_tolerance = 10**(-4)
        while True:
            probability = poisson.pmf(m, max_value * self.time_step)
            sum_poisson += probability
            if sum_poisson >= 1 - error_tolerance:
                break
            m += 1
        return sum_poisson, m + 1

    def create_initial_probability(self):
        # Initialize with the system being completely empty (idle)
        row = np.zeros(self.num_states)
        row[0] = 1  # Start with all probability in the first state
        return row

    def generate_states_with_strict_constraint(self):
        states = []
        for queue_size in range(self.K + 1):
            for phase_distribution in product(range(self.s + 1), repeat=self.k):
                if sum(phase_distribution) > self.s or sum(phase_distribution) + queue_size > self.s + self.K:
                    continue
                if sum(phase_distribution) < self.s and queue_size > 0:
                    continue
                states.append((queue_size,) + tuple(phase_distribution))
        return states

    def create_q_matrix_corrected(self):
        """
        Create the Q matrix for the given system parameters with the corrected transition rules.
        """
        states = self.generate_states_with_strict_constraint()
        Q = np.zeros((self.num_states, self.num_states))
        state_indices = {state: index for index, state in enumerate(states)}

        # Populate the Q matrix
        for state in states:
            state_index = state_indices[state]
            queue_size, *phase_distribution = state

            if self.k == 1:
                # Special case for k = 1
                if queue_size < self.K:
                    if sum(phase_distribution) < self.s:  # Server available
                        new_state = (queue_size, phase_distribution[0] + 1)
                    else:  # Server full, customer goes to queue
                        new_state = (queue_size + 1, phase_distribution[0])
                    new_state_index = state_indices.get(new_state)
                    if new_state_index is not None:
                        Q[state_index, new_state_index] = self.lamb

                if phase_distribution[0] > 0:  # Customer in service
                    if queue_size > 0:  # Queue is not empty
                        new_state = (queue_size - 1, phase_distribution[0])
                    else:  # Queue is empty
                        new_state = (queue_size, phase_distribution[0] - 1)
                    new_state_index = state_indices.get(new_state)
                    if new_state_index is not None:
                        Q[state_index, new_state_index] = self.mu * phase_distribution[0]

            else:
                # General case for k > 1
                if queue_size < self.K:
                    if sum(phase_distribution) < self.s:
                        new_state = (queue_size, phase_distribution[0] + 1) + tuple(phase_distribution[1:])
                    else:
                        new_state = (queue_size + 1,) + tuple(phase_distribution)
                    new_state_index = state_indices.get(new_state)
                    if new_state_index is not None:
                        Q[state_index, new_state_index] = self.lamb

                for phase in range(1, self.k + 1):
                    if phase_distribution[phase - 1] > 0:
                        if phase < self.k:
                            new_phase_dist = list(phase_distribution)
                            new_phase_dist[phase - 1] -= 1
                            new_phase_dist[phase] += 1
                            new_state = (queue_size,) + tuple(new_phase_dist)
                        else:
                            if queue_size > 0:
                                new_state = (queue_size - 1, phase_distribution[0] + 1) + tuple(phase_distribution[1:-1]) + (phase_distribution[-1] - 1,)
                            else:
                                new_state = (queue_size,) + tuple(phase_distribution[:-1]) + (phase_distribution[-1] - 1,)
                        new_state_index = state_indices.get(new_state)
                        if new_state_index is not None:
                            Q[state_index, new_state_index] = self.mu * phase_distribution[phase - 1]

        for i in range(self.num_states):
            Q[i, i] = -np.sum(Q[i, :])

        return Q


    def estimate_transient_probabilities_and_metrics_with_cutoff(
      self,
      time_steps,
      arrival_cutoff_time
  ):
      """
      Computes TEW over [0, T'] while allowing arrivals only over [0, T].

      total_time = T'
      arrival_cutoff_time = T
      """

      self.time_step = self.total_time / time_steps
      num_states = self.calculate_number_of_states()

      phi = self.p_0
      total_waiting_time = 0.0
      cumulative_delayed_fires = 0.0

      original_lambda = self.lamb

      last_step_metrics = {}

      for step in range(time_steps):

          current_time_start = step * self.time_step

          if current_time_start < arrival_cutoff_time:
              self.lamb = original_lambda
          else:
              self.lamb = 0.0

          Q = self.create_q_matrix_corrected()
          max_value = np.abs(np.diag(Q)).max()

          if max_value == 0:
              pi_t = phi.copy()
          else:
              p_tilde = Q / max_value
              p_tilde += np.eye(num_states)

              _, m = self.check_poisson_sum(max_value)

              p = [
                  np.exp(-max_value * self.time_step)
                  * (max_value * self.time_step) ** n
                  / factorial(n)
                  for n in range(m + 1)
              ]

              phi_list = [phi]
              for n in range(1, m + 1):
                  phi_list.append(phi_list[-1].dot(p_tilde))

              pi_t = sum(phi_list[n] * p[n] for n in range(m + 1))
              pi_t = pi_t / np.sum(pi_t)

          phi = pi_t

          E_Lq, E_L, P_W, P_f, Delayed_Fires = self.calculate_metrics(pi_t)

          total_waiting_time += E_Lq * self.time_step
          cumulative_delayed_fires += Delayed_Fires

          if step == time_steps - 1:
              last_step_metrics = {
                  "E_Lq_Tprime": E_Lq,
                  "E_L_Tprime": E_L,
                  "P_W_Tprime": P_W,
                  "P_f_Tprime": P_f,
                  "TEW": total_waiting_time,
                  "total_delayed_fires": cumulative_delayed_fires,
              }

      self.lamb = original_lambda
      return last_step_metrics


    def calculate_metrics(self, transient_probabilities):
        """
        Calculate queueing metrics for the current interval:
        - E[Lq]: Expected number in queue
        - E[L]: Expected number in system
        - P(W): Probability of waiting
        - P(f): Probability of rejection (system full)
        - Delayed_Fires: Expected number of fires experiencing delay during the interval
        """
        E_Lq = 0      # Expected queue length
        E_L = 0       # Expected total in system
        P_W = 0       # Probability of waiting (queue > 0 or all servers busy)
        P_f = 0       # Probability of rejection (queue at capacity)
        P_delay = 0   # Sum of probabilities where queue > 0

        arrival_rate = self.lamb
        interval_length = self.time_step

        states = self.generate_states_with_strict_constraint()

        for state, prob in zip(states, transient_probabilities):
            l = state[0]                # Number in queue
            in_service = np.sum(state[1:])  # Number in all service phases

            E_Lq += l * prob
            E_L += (l + in_service) * prob

            if l > 0 or in_service == self.s:
                P_W += prob

            if l == self.K:
                P_f += prob

            if l > 0:
                P_delay += prob

        # Expected number of delayed fires in this interval
        Delayed_Fires = arrival_rate * P_delay * interval_length

        return E_Lq, E_L, P_W, P_f, Delayed_Fires

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.colors as mcolors

def compute_required_server_heatmap(
    fire_counts,
    service_times_minutes,
    threshold_minutes=30,
    T=12,
    Tprime=24,
    interval_hours=1/12,
    r=2,
    kappa=40,
    s_min=1,
    s_max=10
):
    rows = []

    threshold_hours = threshold_minutes / 60
    time_steps = int(Tprime / interval_hours)

    for n_fires in fire_counts:
        lamb = n_fires / T   # arrival rate per hour

        for service_min in service_times_minutes:
            eta = service_min / 60   # mean service time in hours
            mu = r / eta             # Erlang phase rate

            required_s = None
            best_tew = None

            for s in range(s_min, s_max + 1):
                model = TransientProbClass(
                    s=s,
                    k=r,
                    K=kappa,
                    total_time=Tprime,
                    lamb=lamb,
                    mu=mu
                )

                metrics = model.estimate_transient_probabilities_and_metrics_with_cutoff(
                    time_steps=time_steps,
                    arrival_cutoff_time=T
                )

                tew = metrics["TEW"]
                best_tew = tew

                if tew <= threshold_hours:
                    required_s = s
                    break

            if required_s is None:
                required_s = s_max + 1

            rows.append({
                "fires": n_fires,
                "service_minutes": service_min,
                "lambda": lamb,
                "eta": eta,
                "required_servers": required_s,
                "TEW": best_tew
            })

    return pd.DataFrame(rows)

# Axis values
fire_counts = list(np.arange(1, 13, 1))          # 1 to 12 fires in 12 hours
#fire_counts = list(np.arange(1, 13))
service_times_minutes = [120, 135, 150, 175, 210, 240]
#service_times_minutes = [120, 130, 140, 150, 160, 170, 180, 190, 200, 210, 220, 230, 240]


df_req = compute_required_server_heatmap(
    fire_counts=fire_counts,
    service_times_minutes=service_times_minutes,
    threshold_minutes=30,
    T=12,
    Tprime=24,
    interval_hours=1/12,   # 5-minute intervals
    r=2,
    kappa=40,
    s_min=1,
    s_max=10
)

# Pivot for heatmap
heatmap_df = df_req.pivot(
    index="fires",
    columns="service_minutes",
    values="required_servers"
)

# Reverse y-axis so 12 appears at top
heatmap_df = heatmap_df.sort_index(ascending=False)

unique_vals = sorted(df_req["required_servers"].unique())
colors = sns.color_palette("YlOrRd", n_colors=len(unique_vals))
cmap = mcolors.ListedColormap(colors)

bounds = [v - 0.5 for v in range(min(unique_vals), max(unique_vals) + 2)]
norm = mcolors.BoundaryNorm(bounds, cmap.N)

plt.figure(figsize=(10, 8))

ax = sns.heatmap(
    heatmap_df,
    annot=True,
    fmt="d",
    cmap=cmap,
    norm=norm,
    cbar=False,
    linewidths=0,
    annot_kws={"fontsize": 16}
)

plt.xlabel("Mean Service Time (Minutes)", fontsize=20)
plt.ylabel("Expected Number of Wildfires in 12 Hours", fontsize=20)

plt.tick_params(axis="x", labelsize=16)
plt.tick_params(axis="y", labelsize=16, rotation=0)

plt.tight_layout()

pdf_path = "server_requirement_heatmap_cutoff.pdf"
plt.savefig(pdf_path, format="pdf", dpi=300, bbox_inches="tight")
plt.show()
```
